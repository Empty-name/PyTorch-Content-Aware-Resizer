## Content-Aware Image Resizing (Seam Carving)

```python
import torch
from seam_carving import load_image, show, seam_carve, seam_insert

img = load_image('assets/couple.jpg')
show(img, title=f"Original Image: {img.shape[1]}x{img.shape[0]}")
```

```python
carved_img = seam_carve(img, n_seams=50)
show(carved_img, title="Seam Carving (Narrower)")

expanded_img = seam_insert(img, n_seams=50)
show(expanded_img, title="Seam Insertion (Wider)")
```
